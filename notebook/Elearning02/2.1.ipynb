## Giải thích ngắn gọn các giải thuật huấn luyện cây quyết định sau: ID3, C4.5, C5.0 và CART.

**Algo ID3**: là thuật toán dạng thuận toán dựa vào phương pháp tham lam(greedy) và chỉ được sử dụng dưới dạng categorical kh dùng được ở dạng numeric.  
**Algo C4.5**: là một bản tiền nhiệm của ID3 chung một tác giả mục đích ra mẳt để khắc phục được cái nhược điểm của ID3 đó chính là chỉ làm việc được với dạng categorical  
**Algo C5.0**: là một bản thương mại hoá và độc quyền là bước tiến lớn(turning point) so với C4.5 vì nó mang lại hiệu suất rất đáng kể có thể nhanh hơn tới vài chục lần bằng cách dùng kỹ thuật boosting tạo ra nhiều cây nhỏ hơn và tổng hợp kết quả cho ra hiệu suất và kết quả tốt hơn.   
**Algo Cart**: là thuật toán gần giống với C4.5 nhưng cách xây dựng là theo cây nhị với 2 nhánh chỉ là yes/no và có thể dùng cho bài toán regression luôn. Do xây dụng trên cấu trrúc cây nhị phân nên khả năng tối ưu bộ nhớ và hiệu năng cao. 

## Giải thuật nào được cài đặt trong thư viện sklearn?
