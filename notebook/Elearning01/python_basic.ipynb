## 4.If and Loop(while and for)

```python
#If Else
age = input("Enter your age: ")
if int(age) >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")    

# Loop 
# While Loop 
# Can use countiune so that skipp the current interationm
count = 0
while count <= 10:
    count += 2
    if count == 6:
        continue
    if count == 8:
        break
    print ("Count ", count )
else:
    print("End Count")

# For Loop
for i in "abc":
    print(i)
```
